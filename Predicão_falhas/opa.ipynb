```python
import pandas as pd

# Carregar o arquivo CSV
csvdata = pd.read_csv(r'Data\recommissioned_batteries\battery24.csv')

# Separar os diferentes modos de operação (carga = 1, descarga = -1, neutro = 0)
modo_carga = csvdata[csvdata['mode'] == 1]
modo_descarga = csvdata[csvdata['mode'] == -1]
modo_neutro = csvdata[csvdata['mode'] == 0]

# Função para calcular as estatísticas de temperatura
def calcular_estatisticas_temperatura(df, column='temperature_battery'):
    media = df[column].mean()
    maximo = df[column].max()
    minimo = df[column].min()
    return media, maximo, minimo

# Calcular as estatísticas para cada modo
media_carga, max_carga, min_carga = calcular_estatisticas_temperatura(modo_carga)
media_descarga, max_descarga, min_descarga = calcular_estatisticas_temperatura(modo_descarga)
media_neutro, max_neutro, min_neutro = calcular_estatisticas_temperatura(modo_neutro)

# Criar o vetor com os resultados
estatisticas_temperatura = {
    'Carga': {'Média': media_carga, 'Máximo': max_carga, 'Mínimo': min_carga},
    'Descarga': {'Média': media_descarga, 'Máximo': max_descarga, 'Mínimo': min_descarga},
    'Neutro': {'Média': media_neutro, 'Máximo': max_neutro, 'Mínimo': min_neutro}
}

# Exibir o resultado
print(pd.DataFrame(estatisticas_temperatura))
```
