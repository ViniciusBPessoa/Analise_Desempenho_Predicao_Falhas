```python
# Inicialização das variáveis
Tropa_do_gera = []
qtd_Valesca_poposuda = 0
Matue = Valor_areas_real[0]  # Primeiro modo

# Loop para contar os modos
for modo in range(len(Valor_areas_real)):
    if Matue == Valor_areas_real[modo]:
        qtd_Valesca_poposuda += 1
    else:
        Tropa_do_gera.append([Matue, qtd_Valesca_poposuda])
        Matue = Valor_areas_real[modo]
        qtd_Valesca_poposuda = 1  # Conta o novo modo atual

    if modo == len(Valor_areas_real) - 1:
        Tropa_do_gera.append([Matue, qtd_Valesca_poposuda])
```

```python
Tropa_do_gera
```

```python
auxiliar_posicao = 0
lista_Quadrante = []
lista_final = []
numero_junto = 4

for posicao, valor in enumerate(Tropa_do_gera):
    nova_posicao = valor[1] + auxiliar_posicao
    selected_rows = csvdata.iloc[auxiliar_posicao + 1 : nova_posicao]
    auxiliar_posicao = nova_posicao

    lista_Quadrante.append(selected_rows.copy())

    if (posicao + 1) % numero_junto == 0:
        lista_final.append(lista_Quadrante.copy())
        lista_Quadrante = []
```
